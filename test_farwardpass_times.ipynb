```python
from cube_nn import (
    CubeValueNNFC,
    CubeValueNNConv,
    CubeValueResNet,
    CubeValueTransformer,
    cube_to_tensor_direct,
    cube_to_tensor_one_hot,
    cube_to_tensor_one_hot_similar,
)

from time import time
from cube import Cube
import torch

cube = Cube()
models = [
    CubeValueNNFC(cube_to_tensor_direct, hidden_size=[128, 64]),
    CubeValueNNFC(cube_to_tensor_one_hot, hidden_size=[256, 256, 64]),
    CubeValueNNConv(),
    CubeValueResNet(),
    CubeValueTransformer(),
]

N = 1000

for model in models:
    model.eval()  # Put model in evaluation mode
    cube_to_tensor = model.get_cube_to_tensor()

    start_time = time()
    for _ in range(N):
        tensor = cube_to_tensor(cube)

        with torch.no_grad():
            output = model(tensor)
    end_time = time()
    print(f"{model.__class__.__name__} evaluation time: {end_time - start_time:.6f} seconds")
```
